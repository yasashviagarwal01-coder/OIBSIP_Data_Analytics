## 11. Category Distribution

```python
data["Category"].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Category", data=data)
plt.title("Number of Orders by Category")
plt.show()
```

### Observation

- Office Supplies has the highest number of orders.
- Furniture is the second most ordered category.
- Technology has the lowest number of orders.

### Business Insight

Office Supplies are purchased frequently, making them an important source of regular sales. Maintaining sufficient inventory for this category can help meet customer demand.

## 12. Sales by Category

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Category", y="Sales", data=data, estimator=sum)
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

### Observation

- Technology generates the highest sales revenue.
- Furniture contributes moderate sales.
- Office Supplies have many orders but lower total sales than Technology.

### Business Insight

Technology products contribute significantly to revenue. The company should continue promoting this category while identifying ways to increase Furniture sales.

## 13. Profit by Category

```python
plt.figure(figsize=(8,5))

sns.barplot(x="Category", y="Profit", data=data, estimator=sum)

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()
```

### Observation

- Technology category generates the highest total profit.
- Office Supplies is the second most profitable category.
- Furniture contributes the lowest profit among all categories.

### Business Insight

Technology products are the most profitable for the company. The Furniture category should be analyzed to improve profitability by reducing discounts or optimizing pricing.

## 15. Profit by Region

```python
plt.figure(figsize=(8,5))

sns.barplot(x="Region", y="Profit", data=data, estimator=sum)

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.show()
```

### Observation

### Business Insight

## 16. Monthly Sales Trend

```python
monthly_sales = data.groupby(data["Order Date"].dt.month)["Sales"].sum()

plt.figure(figsize=(10,5))

monthly_sales.plot(marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)

plt.show()
```

### Observation

- Monthly sales fluctuate throughout the year.
- Some months experience higher sales than others.
- This indicates the presence of seasonal demand.

### Business Insight

The company can plan promotions and inventory based on high-demand months to maximize sales and customer satisfaction.

## 17. Top 10 Best Selling Products



```python
top_products = data.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

top_products
```

```python
plt.figure(figsize=(12,6))

sns.barplot(x=top_products.values,
            y=top_products.index)

plt.title("Top 10 Best Selling Products")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")

plt.show()
```

### Observation

- The top 10 products contribute significantly to the overall sales.
- A few products generate much higher sales than the remaining products.
- Sales are concentrated among the best-selling products.

### Business Insight

The company should ensure adequate inventory for the best-selling products and use targeted marketing to further increase their sales.

## 18. Top 10 Most Profitable Products

```python
top_profit = data.groupby("Product Name")["Profit"].sum().sort_values(ascending=False).head(10)

top_profit
```

```python
plt.figure(figsize=(12,6))

sns.barplot(x=top_profit.values,
            y=top_profit.index)

plt.title("Top 10 Most Profitable Products")
plt.xlabel("Total Profit")
plt.ylabel("Product Name")

plt.show()
```

### Observation

- A small number of products contribute the highest profits.
- Profit generation is concentrated among a few products.
- These products are the company's strongest profit drivers.

## 20. Correlation Heatmap

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    data[["Sales", "Profit", "Quantity", "Discount"]].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

- Sales and Profit show a positive correlation.
- Discount negatively impacts Profit.
- Quantity has a weak relationship with Sales and Profit.

### Business Insight

The company should carefully manage discount policies because higher discounts tend to reduce profitability. Increasing sales while maintaining healthy profit margins will improve overall business performance.
